**Q1:** Aik `Student` class banayein jismein `name` aur `roll_number` private variables (`__`) hon. In variables ko access aur modify karne ke liye getter aur setter methods likhein.

```python
# Write your code here for Q1
class Student:
    def __init__(self, name, roll_number):
        self.__name = name
        self.__roll_number = roll_number

    # Getters
    def get_name(self):
        return self.__name

    def get_roll_number(self):
        return self.__roll_number

    # Setters
    def set_name(self, new_name):
        self.__name = new_name

    def set_roll_number(self, new_roll_number):
        self.__roll_number = new_roll_number


# Testing the class
student1 = Student("Ali", 101)

print("Name:", student1.get_name())
print("Roll Number:", student1.get_roll_number())

student1.set_name("Ahmed")
student1.set_roll_number(102)

print("Updated Name:", student1.get_name())
print("Updated Roll Number:", student1.get_roll_number())
```

**Q2:** Aik `BankAccount` class banayein jismein `__balance` private variable ho. Aik `deposit()` aur aik `withdraw()` method banayein, check karein ke withdraw karte waqt balance zero se kam na ho.

```python
# Write your code here for Q2
class BankAccount:
    def __init__(self, balance=0):
        self.__balance = balance

    def deposit(self, amount):
        self.__balance += amount
        return self.__balance

    def withdraw(self, amount):
        if amount > self.__balance:
            print("Insufficient balance")
        else:
            self.__balance -= amount
        return self.__balance


b = BankAccount(20000)
print(b.deposit(500))     # 20500
print(b.withdraw(1090))   # 19410
print(b.withdraw(50000))  # Insufficient balance, stays 19410
```

**Q3:** Aik class banayein `Employee` jismein salary private ho. Agar koi salary ko minus (negative) value set karne ki koshish kare, toh setter method 'Invalid Salary' ka error message show kare.

```python
# Write your code here for Q3
class Employee:
    def __init__(self, salary):
        self.__salary = salary

    def get_salary(self):
        return self.__salary

    def set_salary(self, new_salary):
        if new_salary < 0:
            print("Invalid Salary")
        else:
            self.__salary = new_salary


e = Employee(50000)
print(e.get_salary())
e.set_salary(-1000)   # Invalid Salary, salary unchanged
print(e.get_salary())
e.set_salary(60000)
print(e.get_salary())
```

**Q7:** Multilevel Inheritance ka use karein. Class `Animal` -> Class `Mammal` -> Class `Dog`. Dog class ke object se parent classes ke features ko access karein.

```python
# Write your code here for Q7
class Animal:
    def __init__(self, name):
        self.name = name

    def eat(self):
        print(f"{self.name} is eating")


class Mammal(Animal):
    def walk(self):
        print(f"{self.name} is walking")


class Dog(Mammal):
    def bark(self):
        print(f"{self.name} is barking")


d = Dog("Rex")
d.eat()
d.walk()
d.bark()
```

**Q10:** Hierarchical Inheritance kya hoti hai? Aik parent class `Shape` banayein aur us se do alag child classes `Circle` aur `Square` banakar samjhayein.

```python
# Write your code here for Q10
class Shape:
    def describe(self):
        print("This is a shape")


class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius ** 2


class Square(Shape):
    def __init__(self, side):
        self.side = side

    def area(self):
        return self.side ** 2


circle = Circle(5)
square = Square(4)
circle.describe()
print("Circle area:", circle.area())
square.describe()
print("Square area:", square.area())
```

**Q13:** Operator Overloading kya hoti hai? Do objects ko `+` operator ke zariye add karne ke liye `__add__` magic method ka use kaise karte hain? Code likhein.

```python
# Write your code here for Q13
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)

    def __str__(self):
        return f"({self.x}, {self.y})"


p1 = Point(1, 2)
p2 = Point(3, 4)
p3 = p1 + p2
print(p3)
```

**Q16:** `abc` module ka use karte hue aik abstract class `Computer` banayein jisme abstract method `process()` ho. Is class ka object banane ki koshish karein aur dekhein kya hota hai.

```python
# Write your code here for Q16
from abc import ABC, abstractmethod

class Computer(ABC):
    @abstractmethod
    def process(self):
        pass


c = Computer()
```

**Q18:** Aik abstract class `Car` banayein jisme aik concrete (normal) method `start_engine()` ho aur aik abstract method `drive()` ho. Child class bana kar dono ko test karein.

```python
# Write your code here for Q18
class Car(ABC):
    def start_engine(self):
        print("Engine started")

    @abstractmethod
    def drive(self):
        pass


class SportsCar(Car):
    def drive(self):
        print("Driving fast")


sc = SportsCar()
sc.start_engine()
sc.drive()
```

**Q19:** Abstraction aur Encapsulation ke darmiyan kya wazeh farq (main difference) hai? Dono ke concepts ko short statements mei define karein.

```python
# Write your code here for Q19
Abstraction is about hiding implementation complexity and exposing only what's necessary — the user of a class doesn't need to know how drive() works internally, just that calling it drives the car. Encapsulation is about hiding data and protecting it from unauthorized access or modification, that's what __balance and the getter/setter pattern from Q1–Q3 were doing. In short: abstraction hides logic, encapsulation hides data.
```

**Q20:** Aik real-world scenario (jaise Payment System) design karein jahan `Payment` aik abstract class ho aur `CreditCardPayment` aur `UPIPayment` uski subclasses hon.

```python
# Write your code here for Q20
class Payment(ABC):
    @abstractmethod
    def pay(self, amount):
        pass


class CreditCardPayment(Payment):
    def pay(self, amount):
        print(f"Paid {amount} using Credit Card")


class UPIPayment(Payment):
    def pay(self, amount):
        print(f"Paid {amount} using UPI")


def process_payment(payment_method, amount):
    payment_method.pay(amount)


process_payment(CreditCardPayment(), 500)
process_payment(UPIPayment(), 250)
```
